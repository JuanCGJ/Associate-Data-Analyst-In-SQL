## 📌 Chapter 1 — Summary Statistics

---

### What is Statistics?

The field of statistics — the practice and study of **collecting and analyzing data**.

**Two main branches:**
| Branch | Description |
|--------|-------------|
| **Descriptive / Summary Statistics** | Describing or summarizing data we already have |
| **Inferential Statistics** | Collecting a sample and applying results to the broader population |


### Types of Data

**Numeric Data:**
| Type | Description | Example |
|------|-------------|----------|
| Continuous | Any value in a range | Height, temperature, time |
| Discrete (Interval/Count) | Countable, whole numbers | Number of students, goals scored |

**Categorical Data:**
| Type | Description | Example |
|------|-------------|----------|
| Nominal | Categories with no natural order | Eye color, country |
| Ordinal | Categories with a meaningful order | Survey rating: Strongly Disagree → Strongly Agree |


### Measures of Center

| Measure | Definition | Example |
|---------|------------|---------|
| **Mean** | Sum of all values ÷ count | Mean of 2, 4, 6 = (2+4+6)/3 = **4** |
| **Median** | Middle value when sorted | Median of 3, 5, 8 = **5** |
| **Mode** | Most frequent value | Mode of 2, 2, 3, 4 = **2** |


### Measures of Spread

Describe how far apart data points are — key for understanding variability.

| Measure | Definition |
|---------|------------|
| **Range** | Max − Min |
| **Variance** | Average of squared distances from the mean (avoids cancellation of +/− values) |
| **Standard Deviation** | √Variance — brings units back to original scale |
| **Quartiles** | Divide data into 4 equal parts (Q1, Q2/median, Q3) |
| **IQR** | Q3 − Q1 — spread of the middle 50%, resistant to outliers |

## 📌 Chapter 2 — Probability and Distributions

---

### Probability Basics

```
P(event) = # ways event can happen / total # of possible outcomes
```

**Example:** Coin flip → P(heads) = 1/2 = **50%**


### Conditional Probability

The probability of event A **given** that event B has already occurred:

```
P(A | B) = P(A and B) / P(B)
```

**Key concepts:**
| Concept | Description |
|---------|-------------|
| **Dependent events** | Outcome of one affects the other (e.g., drawing cards without replacement) |
| **Independent events** | Outcomes do not influence each other |
| **Sampling without replacement** | Removing an item changes subsequent probabilities |


### Continuous Distributions

Unlike discrete distributions, continuous distributions model data that can take any value in a range.

| Distribution | Shape | Example |
|-------------|-------|---------|
| **Uniform** | Flat line — every value equally likely | Bus wait time 0–12 min: P(4–7 min) = 3/12 = **25%** |
| **Bimodal** | Two peaks | Book prices (paperback vs hardback) |
| **Normal** | Bell curve | Blood pressure, exam scores |

> 📌 **Key insight:** Probability = **area under the curve** between two values.

## 📌 Chapter 3 — More Distributions & Central Limit Theorem

---

### Binomial Distribution

Probability distribution of the **number of successes** in a sequence of independent binary events.

**Described by:**
- `n` — total number of events
- `p` — probability of success

**Expected value** = n × p  
> Example: Expected heads in 10 coin flips = 10 × 0.5 = **5**

**Applications:** Clinical trial effectiveness, sports betting outcomes  
> ⚠️ **Requires independent events** — if events are dependent, binomial distribution does NOT apply.


### Normal Distribution (Bell Curve)

The most common distribution in nature — symmetric, bell-shaped.

| Parameter | Role |
|-----------|------|
| **μ (mean)** | Center of the curve; also the median and mode |
| **σ (std dev)** | Controls the width/spread of the curve |

**Empirical Rule (68-95-99.7):**
| Range | % of Data |
|-------|-----------|
| μ ± 1σ | ~68% |
| μ ± 2σ | ~95% |
| μ ± 3σ | ~99.7% |

**Skewness** — direction of the tail:
- Positive skew → tail on the right
- Negative skew → tail on the left

**Kurtosis** — sharpness of the peak:
| Type | Kurtosis | Shape |
|------|----------|-------|
| Leptokurtic | Positive | High peak, fat tails (more extremes) |
| Mesokurtic | Normal | Standard bell curve |
| Platykurtic | Negative | Flat peak, thin tails |


### Central Limit Theorem (CLT)

> The sampling distribution of a statistic becomes **closer to normal** as the sample size increases — regardless of the original population's distribution.

**Requirements:**
- Samples must be **random and independent**
- Larger samples → better approximation to normal

**Why it matters:** Allows us to use normal distribution assumptions for hypothesis testing even when the original data isn't normally distributed.


### Poisson Distribution

Models the **number of events occurring in a fixed period of time or space**, when the average rate is known but events happen randomly.

**Poisson process examples:**
- Number of animals adopted from a shelter per week
- Number of customers arriving at a restaurant per hour
- Number of website visits per day

**Questions it answers:**
- P(at least 5 adoptions per week)
- P(exactly 12 arrivals per hour)
- P(less than 200 website visits today)

## 📌 Chapter 4 — Correlation & Hypothesis Testing

---

### Why Hypothesis Testing?

A framework of theories and methods to **compare populations** and make data-driven decisions.

**Real-world examples:**
- Can a price change lead to increased revenue?
- Will changing a website URL increase traffic?
- Is a medication effective for a health condition?


### Null vs. Alternative Hypothesis

Always start by **assuming no difference exists** — this is the null hypothesis.

| Hypothesis | Symbol | Description |
|-----------|--------|-------------|
| **Null (H₀)** | H₀ | No difference / no effect exists |
| **Alternative (H₁)** | H₁ | A difference or effect does exist |

**Example — Vitamin C and birth gender ratio:**
- H₀: No difference in gender birth ratio between women who take or don't take vitamin C
- H₁: More female births occur among women taking vitamin C supplements


### Hypothesis Testing Workflow

1. **Define target populations** — e.g., women taking vs. not taking vitamin C
2. **Develop null and alternative hypotheses**
3. **Collect or access sample data**
4. **Perform statistical tests** on the sample
5. **Draw conclusions** about the population

**Variables:**
| Variable | Axis | Role | Example |
|----------|------|------|---------|
| **Independent** | X | Unaffected by other data | Vitamin C intake |
| **Dependent** | Y | Affected by other data | Birth gender ratio |


### Controlled Experiments

| Component | Description |
|-----------|-------------|
| **Treatment group** | Receives the treatment (e.g., sees the ad) |
| **Control group** | Does not receive the treatment |
| **Randomization** | Random assignment eliminates selection bias → Randomized Controlled Trial (RCT) |
| **Blinding** | Participants don't know which group they're in (receive a placebo) |
| **Double-blind** | Neither participant nor administrator knows → minimizes all bias |

**RCT vs A/B Testing:**
| Method | Groups | Common In |
|--------|--------|-----------|
| **RCT** | Multiple treatment groups | Science, clinical research |
| **A/B Test** | Exactly 2 groups (treatment + control) | Marketing, engineering |


### Pearson Correlation Coefficient (r)

Quantifies the **strength and direction** of a linear relationship between two variables.

```
r ∈ [-1, 1]
```

| Value | Interpretation |
|-------|----------------|
| ±0.99 | Very strong relationship |
| ±0.75 | Strong relationship |
| ±0.56 | Moderate relationship |
| ±0.21 | Weak relationship |
| ±0.04 | No relationship |

**Sign = direction:**
- +0.75 → as X increases, Y increases
- −0.75 → as X increases, Y decreases

> ⚠️ **Correlation ≠ Causation!**  
> Life expectancy and water bottle price may correlate — but it's the **strength of the economy** (a confounding variable) that actually drives both.

**Confounding variable:** An unmeasured variable that affects the relationship between the variables being studied.


### p-value & Significance Level (α)

**p-value:** Probability of obtaining results at least as extreme as observed, **assuming H₀ is true**.

**Significance level (α):** Threshold set *before* data collection to minimize bias. Typical value: **α = 0.05**

**Decision rule:**
```
If p ≤ α → Reject H₀ (result is statistically significant)
If p > α → Fail to reject H₀
```

> A 5% significance level means: 5% chance of wrongly concluding a difference exists when it doesn't.


### Type I & Type II Errors

| Decision | H₀ True | H₀ False |
|----------|---------|----------|
| **Reject H₀** | ❌ Type I Error (False Positive) | ✅ Correct Conclusion |
| **Accept H₀** | ✅ Correct Conclusion | ❌ Type II Error (False Negative) |

- **Type I Error:** Concluding an effect exists when it doesn't (e.g., approving an ineffective drug)
- **Type II Error:** Missing a real effect (e.g., rejecting an effective drug)
