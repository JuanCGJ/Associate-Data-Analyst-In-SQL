## 📌 Chapter 1 — Storytelling with Data

---

### What is Data Storytelling?

> *The practice of building a narrative around data and visualizations to convey meaning in a powerful and compelling way.*

**Why stories matter:**
- Anecdotes → imagination | Stories → **memorable**
- Add value and context
- Capture audience's attention
- Facilitate decision-making
- Drive change

**The 3-Minute Story / Big Idea:**
- Unique point of view, expressed in **one sentence**
- Clear and concise


### The Data Story Formula

```
DATA STORY = Data + Narrative + Visuals
```

| Component | Requirements |
|-----------|-------------|
| **Data** | Relevant, accurate, reliable, actionable insights |
| **Narrative** | Compelling, easy to understand, prioritizes key points, drives change |
| **Visuals** | Simple, engaging, not misleading |

**Narrative elements:**
- **Main point** — central insight, avoid disconnected facts
- **Explanatory context** — understand background and audience, clarify facts
- **Linear sequence** — logical flow from problem to solution


### Tech vs. Non-Tech Approach

> Technical knowledge is a continuum. Data professionals care about their methods, but the **audience likely cares more about results and implications**.

**Example:** Supply chain agents receiving low-accuracy predictions:
- Don't care about statistics
- Care about **their own pain points**

### ADEPT Technique (Explaining Complex Concepts)

| Letter | Technique |
|--------|-----------|
| **A** | Analogy |
| **D** | Diagram |
| **E** | Example |
| **P** | Plain English |
| **T** | Technical definition |


### Narrative Structure (5 Acts)

| Act | Purpose | Example |
|-----|---------|----------|
| **1. Background** | What motivated the analysis? Who is the focus? | Total profit decreased |
| **2. Insight** | What contributed to the problem? (relevant info only) | Chips +20%, Sweets -30% |
| **3. Insight 2** | Supporting evidence, explain the cause further | Most popular chocolate -50% |
| **4. Climax** | Central insight — what happens if nothing changes? | Loss of $10M next year |
| **5. Next Steps** | Potential solutions, proactive course of action | Rebrand chocolate |

**Types of narratives:**
- Change over time, Correlation, Comparison, Clustering

## 📌 Chapter 2 — Preparing to Communicate the Data

---

### The Right Data

> Include enough contextual insights to support the main point **without overloading** with information.

**Rule:** Use the **minimal amount** of information needed to support your story.


### Identifying Your Audience (Personas)

| Persona | Role | Knowledge | Interests | Right Data |
|---------|------|-----------|-----------|------------|
| **Executive** | CEO, investor, director | Fundamentals | Inform decisions | High-level impact |
| **Project Manager** | PM | Intermediate | Project aligns with objectives | Summary metrics, risk |
| **Tech Team** | Collaborator, supervisor | Expert | Replicate/continue project | Methods, code, model details |
| **General Audience** | Customer, other staff | Novice/generalist | General results and impact | Simple visuals, plain language |

**Levels of audience skepticism:**
- Convince yourself → Convince a friend → Convince a skeptic


### Showing Relevant Statistics

| Stat Type | Description | Example |
|-----------|-------------|----------|
| **Absolute** | Raw numbers | 5,000 customers |
| **Relative** | % change or comparison | 20% revenue drop |
| **Ratio** | Quotient of two variables (normalizes) | Revenue per customer |
| **Aggregates** | Totals, mean, median | Mean salary $51,916 vs Median $34,248 |
| **p-value** | Statistical significance | p < 0.05 |

> ⚠️ **Mean can be misleading** when outliers are present — consider median.


### McCandless Method (Presenting Visualizations)

1. **Introduce** the visualization by name
2. **Answer** the audience's anticipated questions
3. **State** the key insights
4. **Help the audience relate** — explain importance and action items


### Choosing Written vs. Oral Format

| Format | Advantages | Disadvantages |
|--------|------------|---------------|
| **Oral** | Relationship with audience, immediate feedback, non-verbal cues | No permanent record, not for long messages |
| **Written** | Permanent record, easy to share, less emotional, good for code | Hard to verify understanding, no immediate feedback |

**Decision factors:** Audience → Content → Requirements → Channel → **Format**

## 📌 Chapter 3 — Structuring Written Reports

---

### Types of Reports

| Type | Content | Length | Structure | Purpose |
|------|---------|--------|-----------|----------|
| **Informational** | Factual information | Short | Flexible | Inform about facts |
| **Analytical** | Analysis + recommendations | Varies | Strict | Data-driven decisions |

### Final vs. Summary Report

| | **Final Report** | **Summary Report** |
|-|-----------------|-------------------|
| **Elements** | Full data analysis, findings, visuals | Key findings, recommendations, visuals |
| **Format** | Long | Short (< 5 pages) |
| **Audience** | Needs details | No need for details |
| **Note** | — | Links to main document |


### Report Structure

**Introduction:**
- Purpose of the analysis
- Contextual information (what changed, why this matters)
- Question of analysis (what problem are we solving?)

**Body:**
- Data used
- Methods applied
- Analysis performed
- Results obtained

**Conclusions:**
- Restate the question
- Summarize important results
- Add recommendations

## 📌 Chapter 4 — Building Compelling Oral Presentations

---

### Planning Before Building Slides

**Three pillars of a presentation:**

| Pillar | Questions to Answer |
|--------|---------------------|
| **Purpose** | Informative? Instructional? Persuasive? |
| **Audience** | Technical colleagues? Managers? Customers? How many? |
| **Message** | What is the ONE central message? |

**Presentation structure:**
1. Introduction
2. Methods, analysis and model outputs
3. Conclusions and takeaways


### Building Presentation Slides

| Element | Best Practice |
|---------|---------------|
| **Slides** | Support the story; one message per slide |
| **Color** | Max 3 colors; good contrast; color-deficiency inclusive (avoid red+green) |
| **Font** | Serif vs. sans-serif depending on context; appropriate size |
| **Text slides** | Less text; use a headline; layering approach |
| **Visualization slides** | One or two full-size graphs; use layering/highlighting; headline if needed |

> ❌ Slide count or timing is a **bad metric** — focus on clarity and one message per slide.


### Delivering the Presentation

**Preparation:**
- Write a script but **don't memorize** — become familiar with content
- **Anticipate follow-up questions**

**Rehearsal:**
- Stand up, use the slides, speak out loud
- Detect distracting verbal patterns: *um, so, like, basically, actually*
- Find linking statements between sections
- Practice Q&A

**During delivery:**
- Talk **to** the audience (not to the screen)
- Manage timing and pace
- Be aware of emotions and body language
- Open for questions — during or at end


### Avoiding Common Presentation Errors

| Mistake | Fix |
|---------|-----|
| Too long | Keep to ~20 minutes + leave time for Q&A |
| No stated purpose | State purpose at the **beginning** |
| All findings at the end | Guide audience — don't save everything for last |
| Low audience engagement | Strong intro, ask questions, reiterate main idea |
| Monotone delivery | Vary speed, volume, and intonation |
| Distracting body language | Use natural gestures that support the message |

**The effective oral presentation checklist:**
- ✅ Talk to the audience
- ✅ Less is more
- ✅ Consistent take-home message
- ✅ Practice
- ✅ Prepare for follow-up questions
