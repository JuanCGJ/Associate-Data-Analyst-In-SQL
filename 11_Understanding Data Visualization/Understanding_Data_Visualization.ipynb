## 📌 Chapter 1 — Visualizing Distributions

---

### Continuous vs. Categorical Variables

| Type | Definition | Examples |
|------|------------|----------|
| **Continuous** | Things you can do arithmetic on | Mass of squirrels, population, salary |
| **Categorical** | Things that can be classified | Province, job title, pass/fail |


### Histograms — When to Use

✅ Use a histogram when:
1. You have a **single continuous variable**
2. You want to ask questions about the **shape of its distribution**

---

### Histogram Properties

**Modality** — How many peaks?

| Modality | Description |
|----------|-------------|
| Unimodal | One peak |
| Bimodal | Two peaks |
| Trimodal | Three peaks |

**Skewness** — Is it symmetric?

| Skewness | Description |
|----------|-------------|
| Left-skewed | Tail on the left, bulk of data on the right |
| Right-skewed | Tail on the right, bulk of data on the left |
| Symmetric | Balanced around the center |

**Kurtosis** — How many extreme values?

| Type | Shape | Extremes |
|------|-------|----------|
| Mesokurtic | Bell-shaped (normal) | Average |
| Leptokurtic | Narrow peak | Many extreme values |
| Platykurtic | Broad peak | Few extreme values |


### Boxplots — When to Use

✅ Use a boxplot when:
1. You have a **continuous variable split by a categorical variable**
2. You want to **compare distributions** across multiple categories

**Why boxplots over histograms for comparison?**

| Approach | Problem |
|----------|----------|
| Each histogram in its own panel | Hard to compare many groups |
| Stack all histograms in one column | Gets too small to read |
| **Boxplot** ✅ | Compare many distributions at once clearly |

## 📌 Chapter 2 — Visualizing Two Variables

---

### Scatter Plot

✅ Use a scatter plot when:
1. You have **two continuous variables**
2. You want to answer questions about the **relationship** between them

**Correlation patterns to look for:**
- Positive, Negative, No correlation
- Strong positive, Weak positive, Weak negative


### Line Plot

✅ Use a line plot when:
1. You have **two continuous variables**
2. You want to show their relationship
3. **Consecutive observations are connected** (e.g., time series)

> Usually (but not always) the x-axis represents **dates or times**.


### Bar Plot

✅ Use a bar plot when:

**Most common:**
1. You have a **categorical variable**
2. You want to show **counts or percentages** for each category

**Occasionally:**
1. You want another numeric score per category and **need to include zero** in the plot


### Dot Plot

✅ Use a dot plot when:
1. You have a **categorical variable**
2. You want to display numeric scores on a **log scale**, or
3. You want to display **multiple numeric scores** for each category

## 📌 Chapter 3 — Color, Shape & Multi-Variable Plots

---

### Beyond X and Y — Additional Dimensions

**For scatter plots (points):**
| Dimension | Use For |
|-----------|----------|
| Color | Categorical or continuous 3rd variable |
| Size | Numeric 3rd variable |
| Transparency | Overlapping data density |
| Shape | Categorical 3rd variable |

**For line plots:**
| Dimension | Use For |
|-----------|----------|
| Color | Different groups/series |
| Thickness | Emphasis or numeric variable |
| Transparency | Background/secondary lines |
| Line type | Distinguish categories (solid, dashed, dotted) |


### Pair Plot

✅ Use a pair plot when:
1. You have **up to ~10 variables** (continuous, categorical, or mixed)
2. You want to see the **distribution of each variable**
3. You want to see the **relationship between every pair** of variables


### Correlation Heatmap

✅ Use a correlation heatmap when:
1. You have **many continuous variables**
2. You want a **simple overview** of how each pair of variables is related


### Parallel Coordinates Plot

✅ Use a parallel coordinates plot when:
1. You have **many continuous variables**
2. You want to **find patterns** across variables, or
3. You want to **visualize clusters** of observations

## 📌 Chapter 4 — 99 Problems But a Plot Ain't One

---

### When to Use Polar Coordinates

> **Almost never.**

✅ Only when you have a variable that is **naturally circular**:
- Time of day (cycles back around)
- Compass direction (0° = 360°)

For everything else, standard Cartesian coordinates (x/y axes) are clearer and less misleading.

## ✅ Quick Reference — Choosing the Right Plot

| Situation | Best Plot |
|-----------|----------|
| 1 continuous variable, shape of distribution | Histogram |
| 1 continuous variable split by category | Boxplot |
| 2 continuous variables, relationship | Scatter plot |
| 2 continuous variables over time | Line plot |
| 1 categorical + counts/percentages | Bar plot |
| 1 categorical + numeric scores (log scale or multiple) | Dot plot |
| Up to 10 mixed variables, all pairwise relationships | Pair plot |
| Many continuous variables, correlation overview | Correlation heatmap |
| Many continuous variables, patterns/clusters | Parallel coordinates |
| Circular variable (time of day, compass) | Polar coordinates |
